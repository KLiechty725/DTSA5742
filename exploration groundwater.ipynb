```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

import seaborn as sns
```

```python
# Load groundwater
gwl = pd.read_parquet(r"data\gwl_daily.parquet")
gwl_monthly = pd.read_parquet(r"data\gwl_monthly.parquet")
```

```python
plt.figure(figsize=(15,6))
plt.plot(gwl['date'], gwl['gwl'], label='Co')
plt.xlabel('Date')
plt.ylabel('Groundwater Level (ft)')
plt.title('395136108210004')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(15,6))
sns.histplot(gwl['gwl'], kde=True, color='blue')
plt.xlabel('Groundwater Level (ft)')
plt.ylabel('Frequency')
plt.title('Distribution of Groundwater Level - 395136108210004')
plt.show()
```
