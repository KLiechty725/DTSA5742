```python
basin_swe = (
    snow_monthly
    .groupby('date')
    .agg(
        basin_swe_index=('swe_z', 'mean'),
        basin_swe_raw=('swe', 'mean'),
        n_stations=('station', 'nunique')
    )
    .reset_index()
)
```

```python
# temporal features with lag
basin_swe['swe_lag1'] = basin_swe['basin_swe_index'].shift(1)
basin_swe['swe_lag2'] = basin_swe['basin_swe_index'].shift(2)

basin_swe['swe_3mo_mean'] = (
    basin_swe['basin_swe_index']
    .rolling(3)
    .mean()
)
```

```python
basin_swe[:5]
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    basin_swe['date'],
    basin_swe['basin_swe_index'],
    linewidth=1.2
)

plt.axhline(
    0,
    linestyle='--',
    label='Normal snowpack'
)

plt.axhline(1, linestyle=':', alpha=0.5)
plt.axhline(-1, linestyle=':', alpha=0.5)

plt.xlabel('Date')
plt.ylabel('Basin SWE anomaly (standard deviations)')
plt.title('Upper Colorado Basin Snowpack Anomaly')

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
